```python
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
nsteps=1000000

def euler_newton(force, times, x0, v0, m,dt):
    x = [x0]
    v = [v0]
    for i in range(nsteps-1):
        #dt = 0.01
        a = force(times[i], x[i], v[i]) / m
        v.append(v[i] + a*dt)
        x.append(x[i] + v[i]*dt)
    return x, v
times,dt = np.linspace(0, 10, nsteps,retstep=True)
def gravity(t, x0, v0):
    output = -m*g
    return output

def f(x0, v0, g, times):
    output = x0 + v0*times - (1/2)*g*times**2
    return output
x0 = 10
v0 = 0
g = 9.8
m = 10
print(f(x0, v0, g, times))
euler_newton(gravity,times,x0,v0,m,dt)
```

```python
x, v = euler_newton(gravity, times, x0, v0, m, dt)
xtimes = x0+v0*times-(1/2)*g*times**2
vtimes = v0 - g*times
#plt.plot(x,v)
plt.plot(times,v)
plt.plot(times,vtimes,linestyle="dashed")
plt.xlabel("time")
plt.ylabel("velocity")
plt.tight_layout()
plt.show()
plt.plot(times,x)
plt.plot(times,xtimes,linestyle="dashed")
plt.xlabel("time")
plt.ylabel("position")
plt.tight_layout()
plt.show()


```
