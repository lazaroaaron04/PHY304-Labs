```python
import numpy as np
import matplotlib.pyplot as plt
import os
```

Terminal velocity occurs when net force equals 0
linear drag
0=-m*g-bv
v=-m*g/b
Quadratic drag
-m*g-c*v*abs(v)
-m*g+c*v^2
v=-squareroot(m*g/c)

```python
nsteps=1000000

def euler_newton(force, times, x0, v0, m,dt,b):
    x = [x0]
    v = [v0]
    for i in range(nsteps-1):
        #dt = 0.01
        a = force(times[i], x[i], v[i]) / m
        v.append(v[i] + a*dt)
        x.append(x[i] + v[i]*dt)
    return x, v
times,dt = np.linspace(0, 10, nsteps,retstep=True)
def gravity(t, x0, v0):
    output = -m*g
    return output

def f(x0, v0, g, times):
    output = x0 + v0*times - (1/2)*g*times**2
    return output
x0 = 10
v0 = 0
g = 9.8
m = 10
def lin_drag(t,x0,v0):
    output = -m*g-b*v0
    return output
m = 5 #kg
b = 5 #kg/s
g = 9.8 #m/s^2
euler_newton(gravity,times,x0,v0,m,dt,b)
def quad_drag(t,x0,v0):
    output = -m*g-c*v0*np.abs(v0)
    return output
c = 1 #kg/m
euler_newton(lin_drag,times,x0,v0,m,dt,c)
x, v = euler_newton(lin_drag,times,x0,v0,m,dt,b)
plt.plot(times,v)
plt.plot((0,10),(-9.8,-9.8),linestyle= "dashed")
plt.xlabel("time (s)")
plt.ylabel("velocity (m/s)")
plt.tight_layout()
plt.show()
plt.plot(times,x)
plt.xlabel("time (s)")
plt.ylabel("position (m)")
plt.title("linear Drag for Position vs. Time")
plt.tight_layout()
plt.show()
euler_newton(quad_drag,times,x0,v0,m,dt,b)
x, v = euler_newton(quad_drag,times,x0,v0,m,dt,b)
plt.plot(times,v)
plt.plot((0,10),(-7,-7), linestyle="dashed")
plt.xlabel("time (s)")
plt.ylabel("velocity (m/s)")
plt.tight_layout()
plt.show()
plt.plot(times,x)
plt.xlabel("time (s)")
plt.ylabel("position (m)")
plt.tight_layout()
plt.show()
```

The terminal velocity is approached much faster in quadratic drad due to the v^2.
